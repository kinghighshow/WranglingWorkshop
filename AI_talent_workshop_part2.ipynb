## Handling missing values
Missing values appear in:
- AssessmentID
- Assessment Score
- Topic

```python
processed_df = merged_df.copy()

# Fill missing assessment information
processed_df['AssessmentID'] = processed_df['AssessmentID'].fillna('A115')

processed_df['Topic'] = processed_df['Topic'].fillna('No Assessment')

# Calculate the mean Assessment Score
mean_score = processed_df['Assessment Score'].mean()

# Round mean to 2 significant figures
mean_score_2sf = float(f"{mean_score:.2g}")

# Replace missing Assessment Score with the 2-significant-figure mean
processed_df['Assessment Score'] = (
    processed_df['Assessment Score'].fillna(mean_score_2sf)
)

print("Mean Assessment Score used:", mean_score_2sf)

print("\nMissing values after handling:")
print(processed_df.isnull().sum())
```

```python
# Convert numeric columns to numeric data types

processed_df['CandidateID'] = pd.to_numeric(
    processed_df['CandidateID'],
    errors='coerce'
)

processed_df['Years of Experience'] = pd.to_numeric(
    processed_df['Years of Experience'],
    errors='coerce'
)

processed_df['Assessment Score'] = pd.to_numeric(
    processed_df['Assessment Score'],
    errors='coerce'
)

# Convert text columns to string
processed_df['Name'] = processed_df['Name'].astype(str)
processed_df['Skills'] = processed_df['Skills'].astype(str)
processed_df['AssessmentID'] = processed_df['AssessmentID'].astype(str)
processed_df['Topic'] = processed_df['Topic'].astype(str)

print(processed_df.dtypes)
```

```python
# Check for completely duplicated rows

duplicate_count = processed_df.duplicated().sum()

print("Number of completely duplicated rows:", duplicate_count)
```

```python
processed_df = processed_df.drop_duplicates()

print("Dataset shape after removing exact duplicates:", processed_df.shape)
```

```python
# Create an Experienced feature

processed_df['Experienced'] = processed_df['Years of Experience'] >= 3

print(
    processed_df[
        ['CandidateID', 'Years of Experience', 'Experienced']
    ].head(10)
)
```

```python
# Create experience categories

def categorize_experience(years):
    if years < 2:
        return 'Junior'
    elif years < 5:
        return 'Mid-Level'
    else:
        return 'Senior'

processed_df['Experience Level'] = (
    processed_df['Years of Experience']
    .apply(categorize_experience)
)

print(
    processed_df[
        ['Years of Experience', 'Experience Level']
    ].drop_duplicates().sort_values('Years of Experience')
)
```

```python
# Create assessment performance categories

def categorize_score(score):
    if score >= 80:
        return 'Excellent'
    elif score >= 70:
        return 'Good'
    elif score >= 50:
        return 'Average'
    else:
        return 'Needs Improvement'

processed_df['Assessment Performance'] = (
    processed_df['Assessment Score']
    .apply(categorize_score)
)

print(
    processed_df[
        ['Assessment Score', 'Assessment Performance']
    ].head(10)
)
```

```python
# Identify highly experienced and high-performing candidates

processed_df['High Potential'] = (
    (processed_df['Years of Experience'] >= 5) &
    (processed_df['Assessment Score'] >= 80)
)

print(
    processed_df[
        [
            'CandidateID',
            'Years of Experience',
            'Assessment Score',
            'High Potential'
        ]
    ].head(10)
)
```

```python
# Calculate correlation

correlation = processed_df[
    ['Years of Experience', 'Assessment Score']
].corr()

print("Correlation Matrix:")
display(correlation)

print(
    f"\nCorrelation between experience and assessment score: "
    f"{correlation.loc['Years of Experience', 'Assessment Score']:.4f}"
)
```
